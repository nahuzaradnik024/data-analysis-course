### Ejercicio 1 — ¿Qué hay adentro?
 
Te acaban de dar acceso a la base de datos `vetcare.db` de un proyecto nuevo. No sabés qué tablas tiene ni cómo están estructuradas. Antes de escribir cualquier consulta, necesitás entender qué tiene adentro.
 
**Consigna:** Sin mirar el script de creación de arriba, conectate a `vetcare.db` y respondé estas preguntas usando solo Python:
 
1. ¿Qué tablas tiene la base de datos?
2. ¿Qué columnas tiene cada tabla y de qué tipo son?

```python
import sqlite3
import pandas as pd
 
conexion = sqlite3.connect("vetcare.db")
 
# 1. Consultamos sqlite_master para descubrir las tablas existentes.
tablas = pd.read_sql("""
    SELECT name
    FROM sqlite_master
    WHERE type = 'table' AND name NOT LIKE 'sqlite_%'
    ORDER BY name
""", conexion)

print("Tablas encontradas:")
display(tablas)

# 2. Para cada tabla, PRAGMA table_info() muestra sus columnas y tipos.
for tabla in tablas["name"]:
    estructura = pd.read_sql(f"PRAGMA table_info({tabla})", conexion)
    print(f"\nEstructura de la tabla: {tabla}")
    display(estructura[["name", "type"]])

conexion.close()
```

> **Por qué importa esto:** en un proyecto real nunca vas a tener el script de creación a mano. Saber cómo explorar una base de datos desconocida es la primera habilidad que necesitás antes de poder analizarla.

### Ejercicio 4 — WHERE desde Python
 
No siempre necesitas traer toda la tabla. Filtrar en la consulta SQL antes de traer los datos es más eficiente que traer todo y filtrar después en Pandas — especialmente cuando la tabla tiene millones de filas.
 
**Consigna:** Escribe una consulta SQL con `WHERE` para traer solo los perros de la tabla `mascotas`. Sin bucles, sin filtros de Pandas después — el filtro va adentro de la consulta SQL.

```python
import sqlite3
import pandas as pd
 
conexion = sqlite3.connect("vetcare.db")
 
# El filtro se hace en SQL, antes de traer los datos a Pandas.
df_perros = pd.read_sql("""
    SELECT *
    FROM mascotas
    WHERE especie = 'perro'
""", conexion)

print(f"Cantidad de perros: {len(df_perros)}")
print(f"Razas de perros: {df_perros['raza'].unique().tolist()}")

display(df_perros)

conexion.close()
```

### Ejercicio 8 — Dos JOINs encadenados
 
Para ver el nombre del dueño junto con los datos de cada consulta médica, necesitas pasar por tres tablas: `consultas` → `mascotas` → `duenos`.
 
**Consigna:** Escribe una consulta que una las tres tablas y devuelva: nombre del dueño, nombre de la mascota, fecha de la consulta, motivo y costo. Ordená el resultado por fecha ascendente.

```python
import sqlite3
import pandas as pd
 
conexion = sqlite3.connect("vetcare.db")
 
df_historial = pd.read_sql(
    """
    SELECT duenos.nombre AS dueno, mascotas.nombre AS mascota,
           consultas.fecha, consultas.motivo, consultas.costo
    FROM consultas
    JOIN mascotas ON consultas.mascota_id = mascotas.id
    JOIN duenos ON mascotas.dueno_id = duenos.id
    ORDER BY consultas.fecha ASC
    """,
    conexion
)
 
print(df_historial)

conexion.close()
```

### Ejercicio 12 — Consultas condicionales con Python
 
En un sistema real, las consultas que hacés no siempre son las mismas — cambian según lo que el usuario pide o el análisis que estás haciendo. Necesitás construir consultas dinámicamente desde Python.
 
**Consigna:** Escribe una función `buscar_mascotas(especie=None, edad_maxima=None, ciudad=None)` que construya y ejecute una consulta SQL según los parámetros que reciba. Si un parámetro es `None`, no se incluye en el filtro. La función debe devolver un DataFrame.

```python
import sqlite3
import pandas as pd
 
def buscar_mascotas(especie=None, edad_maxima=None, ciudad=None):
    conexion = sqlite3.connect("vetcare.db")
    
    necesita_join = ciudad is not None
    consulta = (
        "SELECT m.* FROM mascotas AS m JOIN duenos AS d ON m.dueno_id = d.id"
        if necesita_join else "SELECT * FROM mascotas"
    )
    condiciones = []
    parametros = []

    prefijo = "m." if necesita_join else ""
    if especie is not None:
        condiciones.append(f"{prefijo}especie = ?")
        parametros.append(especie)
    if edad_maxima is not None:
        condiciones.append(f"{prefijo}edad <= ?")
        parametros.append(edad_maxima)
    if ciudad is not None:
        condiciones.append("d.ciudad = ?")
        parametros.append(ciudad)

    if condiciones:
        consulta += " WHERE " + " AND ".join(condiciones)

    df = pd.read_sql(consulta, conexion, params=parametros)
    conexion.close()
    return df
 
# Probar con distintas combinaciones:
print(buscar_mascotas(especie="perro"))
print(buscar_mascotas(edad_maxima=3))
print(buscar_mascotas(especie="gato", ciudad="Buenos Aires"))
print(buscar_mascotas())  # sin filtros → devuelve todas
```

### Ejercicio 13 — Reporte por veterinario
 
La clínica quiere saber cuánto facturó cada veterinario y cuántas consultas atendió. En la próxima clase vas a aprender `GROUP BY` y funciones de agregación en SQL — por ahora, hacelo con las herramientas que ya conocés.
 
**Consigna:** Trae todas las consultas con `pd.read_sql()` y calculá manualmente, usando Python puro (sin `.groupby()` ni `.mean()`), cuántas consultas atendió cada veterinario y cuánto facturó en total. Generá un reporte formateado.

```python
import sqlite3
import pandas as pd
 
conexion = sqlite3.connect("vetcare.db")
 
df = pd.read_sql("SELECT veterinario, costo FROM consultas", conexion)
conexion.close()
 
reporte = {}

for i in range(len(df)):
    veterinario = df["veterinario"].iloc[i]
    costo = df["costo"].iloc[i]
    if veterinario not in reporte:
        reporte[veterinario] = {"consultas": 0, "total": 0.0}
    reporte[veterinario]["consultas"] += 1
    reporte[veterinario]["total"] += costo

print("REPORTE POR VETERINARIO")
print("─" * 40)
for veterinario, datos in sorted(reporte.items()):
    print(f"{veterinario}: {datos['consultas']} consultas — ${datos['total']:,.0f}")
```

> **Para pensar:** este ejercicio te muestra por qué `GROUP BY` existe. Anotá cuántas líneas necesitaste. En la próxima clase vas a resolver lo mismo en dos líneas con SQL.

### Ejercicio 14 — Múltiples consultas, una conexión
 
Abrir y cerrar una conexión para cada consulta es innecesario y lento. En un análisis real, usás una sola conexión para todas las consultas que necesitás en ese script.
 
**Consigna:** Usando una sola conexión a `vetcare.db`, ejecutá estas cuatro consultas y generá un reporte completo de la clínica:
 
1. Total de dueños registrados
2. Distribución de mascotas por especie (cuántos perros, gatos, etc.)
3. La consulta más cara y la más barata
4. Las mascotas que nunca tuvieron una consulta (están en `mascotas` pero no en `consultas`)

```python
import sqlite3
import pandas as pd
 
conexion = sqlite3.connect("vetcare.db")
 
total_duenos = pd.read_sql("SELECT COUNT(*) AS total FROM duenos", conexion)
df_especies = pd.read_sql(
    "SELECT especie, COUNT(*) AS cantidad FROM mascotas GROUP BY especie", conexion
)
df_extremos = pd.read_sql(
    """
    SELECT motivo, costo, veterinario
    FROM consultas
    WHERE costo = (SELECT MAX(costo) FROM consultas)
       OR costo = (SELECT MIN(costo) FROM consultas)
    ORDER BY costo DESC
    """, conexion
)
df_sin_consulta = pd.read_sql(
    """
    SELECT m.nombre, m.especie
    FROM mascotas AS m
    LEFT JOIN consultas AS c ON m.id = c.mascota_id
    WHERE c.id IS NULL
    """, conexion
)

print(f"Dueños registrados: {total_duenos['total'].iloc[0]}")
print("\nMascotas por especie:")
display(df_especies)
print("\nConsulta más cara y más barata:")
display(df_extremos)
print("\nMascotas sin consultas registradas:")
display(df_sin_consulta)
conexion.close()
```
