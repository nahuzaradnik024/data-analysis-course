### Ejercicio 1 — Tu primera comprehension 
 
Una tienda online tiene una lista de precios sin impuesto. Necesitás generar una nueva lista con el precio final, agregando un 19% de impuesto a todos.
 
**Consigna:** Primero resuelve esto con un bucle `for` tradicional, usando una lista vacía y `.append()`. Después escribí la misma solución como list comprehension. Comprueba que ambas dan el mismo resultado.
 

```python
precios = [12000, 8500, 34000, 5200, 19900]

# Versión 1: con bucle for
precios_con_iva_for = []
for precio in precios:
    precios_con_iva_for.append(precio * 1.19)

# Versión 2: la misma operación en una list comprehension
precios_con_iva = [precio * 1.19 for precio in precios]

print(precios_con_iva_for)
print(precios_con_iva)
print("¿Son iguales?", precios_con_iva_for == precios_con_iva)
```

### Ejercicio 2 — Filtrar sin transformar
 
Un sistema de recursos humanos necesita identificar rápidamente a los empleados mayores de 30 años para un reporte de antigüedad.
 
**Consigna:** Usando una list comprehension, genera una nueva lista que contenga solo las edades mayores a 30. No transformes los valores, solo filtralos.

```python
edades = [22, 35, 28, 41, 19, 33, 27, 45, 24, 38]
 
# Conservar la edad sin modificarla, solo si supera los 30
mayores_30 = [edad for edad in edades if edad > 30]
 
print(mayores_30)
# Esperado: [35, 41, 33, 45, 38]
```

### Ejercicio 3 — Limpiar espacios y símbolos
 
Un formulario web devuelve los códigos postales con espacios sobrantes que llegan junto al texto. Antes de guardarlos en la base de datos, hay que limpiarlos.
 
**Consigna:** Para cada código postal de la lista, usa `.strip()` para eliminar los espacios sobrantes. Resuelve esto primero con un bucle, después con una comprehension.
 

```python
codigos_postales = ["  1642", "1900 ", " 5000 ", "1000", "  3400  "]
 
# Versión 1: con bucle; conservamos los códigos como texto
codigos_limpios_for = []
for codigo in codigos_postales:
    codigos_limpios_for.append(codigo.strip())

# Versión 2: con comprehension
codigos_limpios = [codigo.strip() for codigo in codigos_postales]
print("¿Son iguales?", codigos_limpios_for == codigos_limpios)
 
print(codigos_limpios)
# Esperado: ['1642', '1900', '5000', '1000', '3400']
```

### Ejercicio 4 — Combinar transformación y condición
 
Una plataforma de e-learning quiere identificar a los estudiantes que aprobaron (nota mayor o igual a 6) y mostrar su nota redondeada a un decimal con un mensaje de aprobación.
 
**Consigna:** Usando una sola list comprehension, genera una lista de strings con el formato `"Aprobado: 7.5"` solo para las notas que aprobaron. Las notas reprobadas no deben aparecer en el resultado.

```python
notas = [8.3, 5.5, 6.0, 4.2, 9.1, 5.9, 7.8, 6.5, 3.0, 8.0]
 
# La condición selecciona; el f-string construye el mensaje
aprobados = [f"Aprobado: {nota:.1f}" for nota in notas if nota >= 6]
 
for a in aprobados:
    print(a)
```

> **Para pensar:** ¿en qué orden tenés que escribir la transformación y la condición dentro de la comprehension?

**Respuesta:** Se escribe `[transformación for elemento in lista if condición]`. Al ejecutarse, Python toma cada elemento, comprueba la condición y, solo si se cumple, calcula la transformación y la agrega a la lista.

### Ejercicio 5 — Normalizar categorías 
 
Un inventario tiene categorías de productos escritas de forma inconsistente y algunas en mayúsculas, otras con espacios de más, otras con la primera letra en minúscula. Antes de agrupar por categoría, hay que normalizar el texto.
 
**Consigna:** Escribe una función `normalizar_categoria` que reciba un string y devuelva la categoría limpia: sin espacios sobrantes y con la primera letra en mayúscula. Después usá una list comprehension para aplicarla a toda la lista, y otra comprehension para quedarte solo con las categorías únicas (sin duplicados).

```python
categorias_crudas = ["  tecnologia", "MUEBLES", "Tecnologia ", "muebles", " ROPA", "ropa  "]
 
def normalizar_categoria(categoria):
    return categoria.strip().capitalize()

categorias_normalizadas = [normalizar_categoria(categoria) for categoria in categorias_crudas]

# set elimina duplicados; sorted fija un orden alfabético
categorias_unicas = [categoria for categoria in sorted(set(categorias_normalizadas))]
 
print(categorias_normalizadas)
print(categorias_unicas)
```

### Ejercicio 6 — Limpiar precios con formato mixto
 
Un proveedor envía una lista de precios en un formato particular: todos usan coma como separador de miles y punto como separador decimal (formato que verías en una tienda de Estados Unidos). Necesitás convertirlos a números para poder operar con ellos.
 
**Consigna:** Escribe una función `limpiar_precio_usa` que reciba un string en este formato y devuelva un float. Tené en cuenta que algunos valores pueden no ser convertibles y la función debe manejar ese caso sin que el programa se detenga. Después aplica la función a toda la lista con una comprehension, y generá una segunda lista solo con los valores que se pudieron convertir.

```python
precios_usa = ["$1,200.50", "$45.00", "N/A", "$3,890.25", "", "$120.00"]
 
def limpiar_precio_usa(valor):
    texto_limpio = valor.strip().replace("$", "").replace(",", "")
    try:
        return float(texto_limpio)
    except ValueError:
        return None

precios_procesados = [limpiar_precio_usa(valor) for valor in precios_usa]

# Conservar también un posible precio 0.0: solo descartamos None
precios_validos = [precio for precio in precios_procesados if precio is not None]
 
print(precios_procesados)
print(precios_validos)
```

> **Para pensar:** este formato es distinto al que viste en el artículo de la clase (donde el punto era separador de miles). ¿Qué cambia en la lógica de limpieza?

**Respuesta:** En el formato estadounidense se elimina la coma de miles y se conserva el punto decimal: `1,200.50` pasa a `1200.50`. En el formato con punto de miles y coma decimal, se elimina el punto y se reemplaza la coma por un punto: `1.200,50` pasa a `1200.50`.

### Ejercicio 7 — Reporte de clientes
 
Un sistema de CRM exportó una lista de emails de clientes con errores de formato típicos: mayúsculas mezcladas, espacios, y algunos campos vacíos que hay que descartar. Necesitás una lista limpia para una campaña de email marketing.
 
**Consigna:** Usando comprehensions y los métodos de string que conocés, generá una lista de emails limpios cumpliendo estas reglas:
 
- Sin espacios sobrantes
- Todo en minúsculas (los emails no distinguen mayúsculas, pero conviene normalizar)
- Descartar los que estén vacíos después de limpiarlos
- Descartar los que no contengan el símbolo `@` (no son emails válidos)

```python
emails_crudos = [
    "  Ana.Torres@mail.com",
    "LUIS@MAIL.COM  ",
    "   ",
    "clara_diaz@mail.com",
    "sin-arroba.com",
    "",
    "  Pedro@Mail.Com",
]
 
# Primero normalizar; después aplicar los filtros de la consigna
emails_limpios = [email.strip().lower() for email in emails_crudos]
emails_validos = [email for email in emails_limpios if email != "" and "@" in email]
 
print(emails_validos)
# Esperado: ['ana.torres@mail.com', 'luis@mail.com', 'clara_diaz@mail.com', 'pedro@mail.com']
```

> **Pista:** puedes usar el operador `in` para verificar si un carácter está dentro de un string: `"@" in "ana@mail.com"` da `True`.
