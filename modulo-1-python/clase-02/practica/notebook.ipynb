### Ejercicio 1 — Clasificar una transacción

En cualquier sistema de pagos existe una lógica de clasificación automática. Dependiendo del monto, una transacción recibe una categoría distinta que determina cómo se procesa.

**Consigna:** Dado el monto de una transacción, escribí un programa que lo clasifique según estas reglas y lo imprima con un mensaje descriptivo:


- Menos de $1.000: "micro transacción"
- Entre $1.000 y $9.999: "transacción estándar"
- Entre $10.000 y $99.999: "transacción importante"
- $100.000 o más: "transacción de alto valor"

```python
# Punto de partida
monto = 999

# Evaluar los límites de menor a mayor
if monto < 1000:
    categoria = "micro transacción"
elif monto < 10000:
    categoria = "transacción estándar"
elif monto < 100000:
    categoria = "transacción importante"
else:
    categoria = "transacción de alto valor"

# Mostrar el resultado después de elegir una categoría
print(f"Monto: ${monto} — Categoría: {categoria}")
```

### Ejercicio 2 — Recorrer una lista de precios

Una tienda necesita calcular automáticamente el precio final de todos sus productos aplicando el impuesto correspondiente.

**Consigna:** Recorre la lista de precios con un bucle for y para cada precio calculá e imprimí el valor sin impuesto y con un impuesto del 19%. Al final, imprimí también el total acumulado de todos los precios con impuesto.

```python
# Punto de partida
precios = [8500, 32000, 15900, 7200, 54000, 12300]
impuesto = 0.19

# Empezar el acumulador antes de recorrer los precios
total_general = 0

for precio in precios:
    precio_final = precio * (1 + impuesto)
    total_general += precio_final
    print(f"Precio base: ${precio} → Precio final: ${precio_final:.2f}")

# Mostrar el total una vez terminado el bucle
print(f"Total general: ${total_general:.2f}")
```

### Ejercicio 4 — Filtrar con condiciones

Un analista necesita separar los registros de ventas en dos grupos: las que superan la meta mensual y las que no. Es el tipo de operación de filtrado más básica en data analysis.

**Consigna:** Recorre la lista de ventas y separalas en dos listas nuevas: sobre_meta y bajo_meta. Al final imprimí cuántas hay en cada grupo y el total acumulado de cada uno. La meta es $50.000.

```python
# Punto de partida
ventas = [23000, 87000, 45000, 102000, 31500, 68000, 49000, 95000, 12000, 74000]
meta = 50000

sobre_meta = []
bajo_meta = []

# Separar las ventas según si superan la meta
for venta in ventas:
    if venta > meta:
        sobre_meta.append(venta)
    else:
        # Incluye las ventas iguales a la meta: no la superan
        bajo_meta.append(venta)

# Calcular y mostrar los resultados después de recorrer toda la lista
print(f"Ventas que superan la meta: {len(sobre_meta)}")
print(f"Total sobre la meta: ${sum(sobre_meta)}")
print(f"Ventas que no superan la meta: {len(bajo_meta)}")
print(f"Total que no supera la meta: ${sum(bajo_meta)}")
```

### Ejercicio 5 — Reporte por categoría

En un reporte de ventas real necesitas agrupar los datos por algún criterio — categoría, región, vendedor — y calcular métricas por grupo. Antes de usar pandas, este ejercicio te muestra cómo funciona esa lógica por dentro.

**Consigna:** Recorre la lista de productos y construí un diccionario que agrupe el total de ventas (precio × cantidad) por categoría. Al final imprimí un reporte con el total por categoría y cuál fue la categoría con mayor ingreso.

```python
# Punto de partida
productos = [
    {"nombre": "Notebook",    "categoria": "tecnologia", "precio": 320000, "cantidad": 3},
    {"nombre": "Escritorio",  "categoria": "muebles",    "precio": 85000,  "cantidad": 5},
    {"nombre": "Mouse",       "categoria": "tecnologia", "precio": 8500,   "cantidad": 20},
    {"nombre": "Silla",       "categoria": "muebles",    "precio": 45000,  "cantidad": 8},
    {"nombre": "Monitor",     "categoria": "tecnologia", "precio": 210000, "cantidad": 4},
    {"nombre": "Lámpara",     "categoria": "muebles",    "precio": 12000,  "cantidad": 15},
    {"nombre": "Auriculares", "categoria": "tecnologia", "precio": 32000,  "cantidad": 11},
]

totales_por_categoria = {}

# Acumular los ingresos de cada producto en su categoría
for producto in productos:
    categoria = producto["categoria"]
    ingreso = producto["precio"] * producto["cantidad"]

    if categoria not in totales_por_categoria:
        totales_por_categoria[categoria] = 0

    totales_por_categoria[categoria] += ingreso

# Mostrar el total de cada categoría
print("REPORTE POR CATEGORÍA")
for categoria, total in totales_por_categoria.items():
    print(f"{categoria}: ${total:,}")

# Comparar los ingresos para encontrar la categoría ganadora
if totales_por_categoria:
    categoria_ganadora = max(totales_por_categoria, key=totales_por_categoria.get)
    mayor_ingreso = totales_por_categoria[categoria_ganadora]
    print(f"Mayor ingreso: {categoria_ganadora} (${mayor_ingreso:,})")
else:
    print("No hay productos para analizar")
```
